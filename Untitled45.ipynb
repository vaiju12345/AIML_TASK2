```python
import pandas as pd
data={
    'ID': [1, 2, 3, 4, 5, 5, 7, 8],
    'Name': ['Ram', 'Priya', 'John', 'Sam', 'Anu', 'Anu', 'David', 'Ravi'],
    'Department': ['IT', 'HR', 'IT', 'HR', 'Sales', 'Sales', 'IT', 'HR'],
    'Age': [22, 24, None, 25, 23, 23, None, 26],
    'Salary': [20000, 25000, 30000, None, 28000, 28000, 35000, 22000],
    'City': ['Chennai', 'Bangalore', 'Chennai', 'Mumbai', 'Chennai', 'Chennai', 'Delhi', 'Mumbai']

}
df=pd.DataFrame(data)
```

1.Duplicate Handling

```python
print(df.duplicated())
print(df.drop_duplicates())
```

2.Missing Values

```python
print(df.isnull().sum())

df['Age'] = df['Age'].fillna(df['Age'].mean())

df['Salary'] = df['Salary'].fillna(df['Salary'].mean())

print(df)
```

3. Pivot

```python
result = df.drop_duplicates(subset=['ID']).pivot(
    index='ID',
    columns='Department',
    values='Salary'
)

print(result)
```

4.Melt

```python
result = pd.melt(
    df,
    id_vars=['ID', 'Name'],
    value_vars=['Age', 'Salary'],
    var_name='Details',
    value_name='Value'
)

print(result)
```

5. Pivot Table

```python
result = pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    aggfunc='mean'
)

print(result)
```

6. Merge

```python
department = df[['ID', 'Name', 'Department']]

salary = df[['ID', 'Salary']]

result = pd.merge(
    department,
    salary,
    on='ID',
    how='inner'
)

print(result)
```
